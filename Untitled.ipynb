## Checking Outliers (part of EDA) 

```python
def plotting(col_name, nums):
    plt.subplot(2, 3, nums)
    sns.histplot(df[col_name], kde=True)
    plt.title(col_name)

plt.figure(figsize=(15, 8))

plotting('Age', 1)
plotting('RestingBP', 2)
plotting('Cholesterol', 3)
plotting('FastingBS', 4)
plotting('MaxHR', 5)
plotting('Oldpeak', 6)

plt.tight_layout()
plt.show() 
```

```python
df['RestingBP']=df['RestingBP'].replace(0,df['RestingBP'].median())  # since the data is skewed hence median>mean (better)
df['Cholesterol']=df['Cholesterol'].replace(0,df['Cholesterol'].median())
df['Cholesterol']=df['Cholesterol'].replace(0,df['Cholesterol'].median())
```

```python
def plotting(col_name, nums):
    plt.subplot(2, 3, nums)
    sns.histplot(df[col_name], kde=True)
    plt.title(col_name)

plt.figure(figsize=(15, 8))

plotting('Age', 1)
plotting('RestingBP', 2)
plotting('Cholesterol', 3)
plotting('FastingBS', 4)
plotting('MaxHR', 5)
plotting('Oldpeak', 6)

plt.tight_layout()
plt.show() 
```

```python
def figure(col_name, num):
    plt.subplot(2, 2, num)
    sns.countplot(x=df[col_name], hue=df['HeartDisease'])
    plt.title(f'{col_name} vs Heart Disease')


plt.figure(figsize=(8, 8))

figure('Sex', 1)
figure('ChestPainType', 2)
figure('RestingECG', 3)
figure('ExerciseAngina', 4)

plt.tight_layout()
plt.show()
```
